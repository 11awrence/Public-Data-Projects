# Vimeo link clean-up

```python
# dependencies
import re
import numpy as np
import pandas as pd
```

```python
# path and import 
vimeo_path = "C:/Users/ywa/OneDrive - Stifterverband/Dokumente/Data/Moodle Query/Nov2026/VimeoTable.csv"

vimeo_df = pd.read_csv(vimeo_path, sep = ",")
print(vimeo_df.columns)

vimeo_sub = vimeo_df[["vimeo url"]].head(200)

vimeo_sub.head()
```

```python
# create column youtube if contains string
vimeo_sub["source"] = ["YouTube" if "youtube" in str(x) else "Vimeo" for x in vimeo_sub["vimeo url"]]

vimeo_sub.head()
```

```python
# clean the vimeo strings
regex_id = r"/(\d{7,9})"

# extract video ids
vimeo_sub["id"] = np.where(vimeo_sub["source"] == "Vimeo", vimeo_sub["vimeo url"].str.extract(regex_id, expand = False), None)

vimeo_sub
```
